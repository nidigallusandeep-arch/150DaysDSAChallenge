# 97 What is a Linked List?
- A Linked List is a linear data structure where each element is called a node.

Each node contains two things:
- Data – the value.
- Next – a reference to the next node.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
first = Node(10)
second = Node(20)
third = Node(30)
first.next = second
second.next = third
print(first.data)
print(first.next.data)
print(first.next.next.data)
```
