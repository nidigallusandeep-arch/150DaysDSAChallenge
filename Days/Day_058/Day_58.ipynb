# 88.Stack basics

```python
class Solution:
    def Stack_basics(self):
        stack=[]
        stack.append(10)
        stack.append(20)
        stack.append(30)
        print("stack:",stack)
        print("Top element:",stack[-1])
        removed=stack.pop()
        print("Removed element:",removed)
        print("stack after pop:",stack)
        return stack
obj=Solution()
obj.Stack_basics()
```
