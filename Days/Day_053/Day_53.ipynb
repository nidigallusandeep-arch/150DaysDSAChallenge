# 84 Intersection using HashSet
- arr1 = [1, 2, 3, 4, 5]
- arr2 = [3, 4, 5, 6, 7]
output:-[3, 4, 5]

```python
class Solution:
    def intersection(self,arr1,arr2):
        set1=set(arr1)
        result=[]
        for num in arr2:
            if num in set1:
                result.append(num)
        return result
arr1=[1,2,3,4,5]
arr2=[3,4,5,6,7]
obj=Solution()
print(obj.intersection(arr1,arr2))
```
