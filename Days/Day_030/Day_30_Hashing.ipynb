# 57.All Triplets with Zero Sum
- Input: arr[] = [0, -1, 2, -3, 1]
- Output: [[0, 1, 4], [2, 3, 4]]
- Explanation: Triplets with sum 0 are:
arr[0] + arr[1] + arr[4] = 0 + (-1) + 1 = 0
arr[2] + arr[3] + arr[4] = 2 + (-3) + 1 = 0

```python
class Solution:
    def findTriplets(self, arr):
        result = []
        n = len(arr)
        for i in range(n):
            for j in range(i + 1, n):
                for k in range(j + 1, n):
                    if arr[i] + arr[j] + arr[k] == 0:
                        result.append([i, j, k])
        return result
arr = [0, -1, 2, -3, 1]
obj = Solution()
print(obj.findTriplets(arr))
```
