```python
class Solution:
    def multiply(self, A, B):
        result = [[0 for j in range(len(B[0]))]
                  for i in range(len(A))]
        for i in range(len(A)):
            for j in range(len(B[0])):
                for k in range(len(B)):
                    result[i][j] += A[i][k] * B[k][j]
        return result
A = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
B = [[10, 11, 12],
     [13, 14, 15],
     [16, 17, 18]]
obj = Solution()
result = obj.multiply(A, B)
for row in result:
    print(row)
```
