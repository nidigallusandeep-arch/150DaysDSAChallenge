# Count Sum Pairs in Sorted
- Input: arr[] = [-1, 1, 5, 5, 7], target = 6
- Output: 2
- Explanation: There are 3 pairs which sum up to 6 : {1, 5}, {1, 5} and {-1, 7}.

```python
class Solution:
    def countPairs(self, arr, target):
        left = 0
        right = len(arr) - 1
        count = 0
        while left < right:
            total = arr[left] + arr[right]
            if total == target:
                count += 1
                left += 1
                right -= 1
            elif total < target:
                left += 1
            else:
                right -= 1
        return count
arr=arr = [-1, 1, 5, 5, 7]
target = 6
obj=Solution()
print(obj.countPairs(arr,target))
```
