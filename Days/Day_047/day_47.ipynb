# 78.Closest Pair Sum
- Input: arr[] = [10, 30, 20, 5], target = 25
- Output: [5, 20]
- Explanation: As 5 + 20 = 25 is closest to 25.

```python
class Solution:
    def sumClosest(self, arr, target):
        if len(arr) < 2:
            return []
        arr.sort()
        left = 0
        right = len(arr) - 1
        best_pair = []
        min_diff = float('inf')
        while left < right:
            current_sum = arr[left] + arr[right]
            diff = abs(current_sum - target)
            if diff < min_diff:
                min_diff = diff
                best_pair = [arr[left], arr[right]]
            if current_sum < target:
                left += 1
            elif current_sum > target:
                right -= 1
            else:
                break
        return best_pair
arr=[10,30,20,5]
target=25
obj=Solution()
print(obj.sumClosest(arr,target))
```
