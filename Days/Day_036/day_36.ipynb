# 63.remove_dupulicate

```python
class Solution:
    def remove_dupulicate(self,arr):
        result=[]
        for num in arr:
            if num not in result:
                result.append(num)
        return result
arr=[10,10,20,30,40,30,40,50]
obj=Solution()
print(obj.remove_dupulicate(arr))
```
