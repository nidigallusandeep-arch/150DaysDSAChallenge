```python
# 43.Count Even Numbers
```

```python
numbers = [10, 15, 20, 25, 30, 35]

count = 0

for n in numbers:
    if n % 2 == 0:
        count += 1

print(count)
```

```python
# 44. Count Odd Numbers
numbers = [10, 15, 20, 25, 30, 35,47]

count = 0

for n in numbers:
    if n % 2 != 0:
        count += 1

print(count)
```
