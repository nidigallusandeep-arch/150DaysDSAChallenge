# 91.Reverse Words in a String III
- Example
- Input: s = "Mr Ding"
- Output: "rM gniD"

```python
class Solution(object):
    def reverseWords(self, s):
        words=s.split()
        result=[]
        for word in words:
            result.append(word[::-1])
        return " ".join(result)

s="Mr Ding"
obj=Solution()
print(obj.reverseWords(s))
```
