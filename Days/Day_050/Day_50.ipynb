# First unique element
- input=arr = [4, 5, 1, 2, 1, 5, 4]
- output=2

```python
class Solution:
    def firstunique(self,arr):
        frequency={}
        for num in arr:
            frequency[num]=frequency.get(num,0)+1
        for num in arr:
            if frequency[num]==1:
                return num
        return -1
arr=[4,5,1,2,1,5,4]
obj=Solution()
print(obj.firstunique(arr))
```
