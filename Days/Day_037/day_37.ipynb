# 64.two_sum

```python
class Solution:
    def two_sum(self,arr,target):
        seen={}
        for i in range(len(arr)):
            required=target-arr[i]
            if required in seen:
                return [seen[required],i]
            seen[arr[i]]=i
        return []
arr=[2,3,4,7,11,14,13,32]
target=9
obj=Solution()
print(obj.two_sum(arr,target))
```
