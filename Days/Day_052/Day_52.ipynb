# 83.Union of arrays
- arr1 = [1, 2, 3, 4]
- arr2 = [3, 4, 5, 6]

- output:[1, 2, 3, 4, 5, 6]

```python
class Solution:
    def find_union(self,arr1,arr2):
        result=set()
        for num in arr1:
            result.add(num)
        for num in arr2:
            result.add(num)
        return list(result)
arr1=[1,2,3,4]
arr2=[3,4,5,6]
obj=Solution()
print(obj.find_union(arr1,arr2))
```
