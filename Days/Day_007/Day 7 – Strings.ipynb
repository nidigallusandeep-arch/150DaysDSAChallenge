# 19 Search Pattern
- Given two strings, a text string txt and a pattern string pat, both consisting of lowercase English alphabets. Return the starting indices (0-based) of all the occurrences of the pattern string pat in the text string txt.

-Input: txt = "geeksforgeeks", pat = "geek"
- Output: [0, 8]
- Explanation: The string "geek" occurs twice in txt, one starts at index 0 and the other at index 8.


```python
class Solution:
    def search(self,pat,txt):
        result=[]
        for i in range(len(txt)-len(pat)+1):
            if txt[i:i+len(pat)]==pat:
                result.append(i)
        return result
txt = "geeksforgeeks"
pat="geek"
obj=Solution()
print(obj.search(pat,txt))
```

# 20.Min Chars to Add for Palindrom
- Given a string s, the task is to find the minimum characters to be added at the front to make the string palindrome.
- Input: s = "abc"
- Output: 2
- Explanation: Add 'b' and 'c' at front of the above string to make it palindrome: "cbabc"

```python
class Solution:
    def minChar(self, s):
        count = 0

        while s != s[::-1]:
            s = s[:-1]
            count += 1

        return count
s="abc"
obj=Solution()
print(obj.minChar(s))
```
