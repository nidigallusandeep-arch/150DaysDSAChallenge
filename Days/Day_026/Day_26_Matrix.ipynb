# 51. Main Diagonal
- 1 2 3
- 4 5 6
- 7 8 9

output:- 3 5 7

```python
class Solution:
    def mainDiagonal(self, arr):
        for i in range(len(arr)):
            print(arr[i][i], end=" ")
arr = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

obj = Solution()
print(obj.mainDiagonal(arr))
```

# 52.Secondary Diagonal

arr = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]
