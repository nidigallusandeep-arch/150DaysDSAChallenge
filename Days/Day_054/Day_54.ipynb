# 85 Longest Consecutive Sequence
- arr = [100, 4, 200, 1, 3, 2]
- output:4

```python
class Solution:
    def longestConsecutiove(self,arr):
        nums=set(arr)
        logest=0
        for num in nums:
            if num-1 not in nums:
                current=num
                count=1
            while current+1 in nums:
                current+=1
                count+=1
            logest=max(logest,count)
        return logest
arr=[100,4,200,1,3,2]
obj=Solution()
print(obj.longestConsecutiove(arr))
```
