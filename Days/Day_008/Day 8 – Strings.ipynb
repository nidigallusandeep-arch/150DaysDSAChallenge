# 21.Strings Rotations of Each Othe
- You are given two strings s1 and s2, of equal lengths. The task is to check if s2 is a rotated version of the string s1.

- Input: s1 = "abcd", s2 = "cdab"
- Output: true
- Explanation: After 2 right rotations, s1 will become equal to s2.


```python
class Solution:
    def areRotations(self, s1, s2):
        
        # Check if both strings have same length
        if len(s1) != len(s2):
            return False
        
        # Join s1 with itself and check s2
        if s2 in (s1 + s1):
            return True
        
        return False
s1="abcd"
s2="cdab"
obj=Solution()
print(obj.areRotations(s1,s2))
```

```python
# 22.6. Remove Duplicate Characters
class Solution:
    def removeDuplicates(self, s):
        result = ""

        for ch in s:
            if ch not in result:
                result += ch

        return result

s = "programming"

obj = Solution()
print(obj.removeDuplicates(s))
```
