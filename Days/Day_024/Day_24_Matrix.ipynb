# 46.Find Maximum elements
- arr = [
    [10, 25, 3],
    [45, 5, 16],
    [7, 8, 90]]
- output:90

```python
class Solution:
    def maximum(self,arr):
        maximum=arr[0][0]
        for row in arr:
            for value in row:
                if value >maximum:
                    maximum=value
        return maximum
arr=[[10,25,3],
    [45,5,16],
    [7,8,90]]
obj=Solution()
print(obj.maximum(arr))
```

# 47.Find minmum element
- arr = [
    [10, 25, 3],
    [45, 5, 16],
    [7, 8, 90]]
- output=3

```python
class Solution:
    def minmum(self,arr):
        minimum=arr[0][0]
        for row in arr:
            for value in row:
                if value<minimum:
                    minimum=value
        return minimum
arr=[[10, 25, 3],
     [45, 5, 16],
     [7, 8, 90]]
obj=Solution()
print(obj.minmum(arr))
```

# 48. Print Row-wise
- arr = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

- output:--

- Row 0 : [[1, 2, 3]
- Row 1 : [4, 5, 6]
- Row 2 : [7, 8, 9]]


```python
class Solution:
    def printRows(self,arr):
        for i in range(len(arr)):
            print("Row", i,":",arr[i])
arr=[[1,2,3],
    [4,5,6],
    [7,8,9]]
obj=Solution()
print(obj.printRows(arr))
```
