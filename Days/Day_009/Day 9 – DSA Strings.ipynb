# 23  Reverse a String

```python
class Solution:
    def reverse_string(self,arr):
        return arr[::-1]
obj=Solution()
print(obj.reverse_string("Sandeep"))
```

# 24 Check Palindrome
- Question: "madam" palindrome aa?

```python
class Solution:
    def palindrome(self,arr):
        return arr==arr[::-1]
obj=Solution()
print(obj.palindrome("MadaM"))
```

# 25.Count Vowels

```python
class Solution:
    def count_vowels(self,s):
        count=0
        for ch in s.lower():
            if ch in "aeiou":
                count +=1
        return count
obj=Solution()
print(obj.count_vowels("sandeep"))
```

# 26. Count Frequency of Each Character

```python
class Solution:
    def frequency(self,s):
        for ch in set(s):
            print(ch," : ",s.count(ch))
obj=Solution()
print(obj.frequency("banana"))
```

# 27. Remove Spaces

```python
class Solution:
    def remove_spaces(self,s):
        return s.replace(" ","")
obj=Solution()
print(obj.remove_spaces("Hello World"))
```

# 28 Count Words

```python
class Solution:
    def count_words(self,s):
        return len(s.split())
obj=Solution()
print(obj.count_words("I am learning DSA"))
```
