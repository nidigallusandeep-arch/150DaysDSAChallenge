# 15Add Binary Strings
- Given two binary strings s1 and s2 consisting of only 0s and 1s. Find the resultant string after adding the two Binary Strings.

- Input: s1 = "1101", s2 = "111"
- Output: 10100
- Explanation:
 1101
+ 111
10100

```python
class Solution:
    def addBinary(self,s1,s2):
        num1=int(s1,2)
        num2=int(s2,2)
        total=num1+num2
        return bin(total)[2:]
s1="1101"
s2="111"
obj=Solution()
print(obj.addBinary(s1,s2))
```

```python
class Solution:
    def areAnagrams(self, s1, s2):
        if len(s1) != len(s2):
            return False

        return sorted(s1) == sorted(s2)

s1 = "geeks"
s2 = "kseeg"
obj = Solution()
print(obj.areAnagrams(s1, s2))
```
