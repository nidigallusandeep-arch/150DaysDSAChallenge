# 66.Compare First and Last Elements

```python
class Solution:
    def compareFirstLast(self, arr):
        left = 0
        right = len(arr) - 1
        while left < right:
            print("First:", arr[left], "Last:", arr[right])
            left += 1
            right -= 1
s = Solution()
arr = [10, 20, 30, 40, 50]
s.compareFirstLast(arr)   
```
