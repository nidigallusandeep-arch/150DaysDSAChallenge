# 67.Count even numbers

```python
class Solution:
    def count_even(self,arr):
        count=0
        for i in arr:
            if i%2==0:
                count+=1
        return count
arr=[11,12,13,14,15,16,17,18]
obj=Solution()
print(obj.count_even(arr))
```

# 68 Count odd numbers

```python
class Solution:
    def count_odd(self,arr):
        count=0
        for num in arr:
            if num %2!=0:
                count+=1
        return count
arr=[11,12,13,14,15,16]
obj=Solution()
print(obj.count_odd(arr))
```
