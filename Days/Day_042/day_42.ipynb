# 71.Find Missing Number
- Input:  [1, 2, 3, 5, 6]
- Output: 4

```python
class Solution:
    def missing_number(self,arr):
        n=len(arr)+1
        expected = n * (n + 1) // 2
        actual = sum(arr)
        return expected - actual
arr=[1,2,3,5,6]
obj=Solution()
print(obj.missing_number(arr))
```

# Logic
- For 1 to n:
- Sum = n × (n + 1) / 2
- Expected sum:
- 1 + 2 + 3 + 4 + 5 + 6 = 21
- Actual : 1 + 2 + 3 + 5 + 6 = 17
- Missing
- 21 - 17 = 4
