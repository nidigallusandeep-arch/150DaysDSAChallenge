# 29.Number of Occurrenc
- Given a sorted array arr[] and a number target, find the number of occurrences of target in given arra

-Input: arr[] = [1, 1, 2, 2, 2, 2, 3], target = 2
- Output: 4
- Explanation: target = 2 occurs 4 times in the given array so the output is 4.

```python
class SOlution:
    def countFreq(self,arr,target):
        count=0
        for num in arr:
            if num==target:
                count+=1
        return count
arr=[1,1,2,2,2,2,3]
target=2
obj=SOlution()
print(obj.countFreq(arr,target))
```

# 30.Sorted and Rotated Minimum
- A sorted array of distinct elements arr[] is rotated at some unknown point, the task is to find the minimum element in it. 

- Input: arr[] = [5, 6, 1, 2, 3, 4]
- Output: 1
- Explanation: 1 is the minimum element in the array

```python
class Solution:
    def find_min(self,arr):
        return min(arr)
arr=[5,6,1,2,3,4]
obj=Solution()
print(obj.find_min(arr))
```

# 31.Search in Rotated Sorted Arra
- Given an array arr[] of distinct elements, which was initially sorted in ascending order but then rotated at some unknown pivot, the task is to find the index of a target key.  If the key is not present in the array, return -1.

- Input: arr[] = [5, 6, 7, 8, 9, 10, 1, 2, 3], key = 3
- Output: 8
- Explanation: 3 is found at index 8.

```python
class Solution:
    def search(self,arr,key):
        for i in range(len(arr)):
            if arr[i]==key:
                return i
        return -1
arr=[5, 6, 7, 8, 9, 10, 1, 2, 3]
key=3
obj=Solution()
print(obj.search(arr,key))
```
