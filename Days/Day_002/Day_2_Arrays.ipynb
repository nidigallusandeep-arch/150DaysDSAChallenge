# 5.Rotate Array
- Input: arr[] = [1, 2, 3, 4, 5], d = 2
- Output: [3, 4, 5, 1, 2]

```python
class Solution:
    def roratearray(self,arr,d):
        d=d%len(arr)
        arr[:]=arr[d:]+arr[:d]
        return arr
arr=[1,2,3,4,5]
d=2
obj=Solution()
print(obj.roratearray(arr,d))
```

# 6.Next Permutation
- Input: arr[] = [2, 4, 1, 7, 5, 0]
- Output: [2, 4, 5, 0, 1, 7]

```python
class Solution:
    def nextpermutation(self,arr):
        n=len(arr)
        i=n-2
        while i>=0 and arr[i]>=arr[i+1]:
            i-=1
        if i>=0:
            j=n-1
            while arr[j]<=arr[i]:
                j-=1
            arr[i],arr[j]=arr[j],arr[i]
        arr[i+1:]=reversed(arr[i+1:])
        return arr
arr=[2,3,1,7,5,0]
obj=Solution()
print(obj.nextpermutation(arr))
```

# 7.Majority Element – More Than n/3
- Input: arr[] = [2, 2, 3, 1, 3, 2, 1, 1]
- Output: [1, 2]

```python
class Solution:
    def findMajority(self, arr):
        count = {}
        result = []

        for num in arr:
            count[num] = count.get(num, 0) + 1

        for num in count:
            if count[num] > len(arr) // 3:
                result.append(num)

        return sorted(result)


arr = [2, 2, 3, 1, 3, 2, 1, 1]

obj = Solution()
print(obj.findMajority(arr))
```

# 8.Stock Buy and Sell – Multiple Transaction Allowed
- Input: prices[] = [100, 180, 260, 310, 40, 535, 695]
- Output: 865

```python
class Solution:
    def maxProfit(self, prices):
        profit=0
        for i in range(1,len(prices)):
            if prices[i]>prices[i-1]:
                profit+=prices[i]-prices[i-1]
        return profit
arr=[100, 180, 260, 310, 40, 535, 695]
obj=Solution()
print(obj.maxProfit(arr))
```
