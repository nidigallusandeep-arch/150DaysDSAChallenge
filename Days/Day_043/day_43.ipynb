# 72 Find Common Elements

- A = [1, 2, 3, 4]
- B = [3, 4, 5, 6]
- Output: [3, 4]

```python
class Solution:
    def common_element(self,arr1,arr2):
        result=[]
        for num in arr1:
            if num in arr2 and num not in result:
                result.append(num)
        return result
arr1=[1,2,3,4]
arr2=[3,4,5,6]
obj=Solution()
print(obj.common_element(arr1,arr2))
```

# 18. Find Maximum Difference 
- maximum element - minimum element

- Input:  [10, 5, 20, 8, 15]
- Output: 15

```python
class Solution:
    def maximum_difference(self,arr):
        minimum=arr[0]
        maximum=arr[0]
        for num in arr:
            if num < minimum:
                minimum=num
            if num > maximum:
                maximum=num
        return maximum-minimum
arr=[10,5,20,8,15]
obj=Solution()
print(obj.maximum_difference(arr))
```
