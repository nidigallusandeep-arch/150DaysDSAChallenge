# 74.Rotate Array Left
- Rotate by 2 positions
- Input:  [1, 2, 3, 4, 5]
- Output: [3, 4, 5, 1, 2]

```python
class Solution:
    def rotate_array(self,arr,k):
        n=len(arr)
        k=k%n
        return arr[k:]+arr[:k]
arr=[1,2,3,4,5]
obj=Solution()
print(obj.rotate_array(arr,2))
```

# 75. Rotate Array Right
- Rotate by 2 positions
- Input:  [1, 2, 3, 4, 5]
- Output: [4, 5, 1, 2, 3]

```python
class Solution:
    def rotate_array(self,arr,k):
        n=len(arr)
        k=k%n
        return arr[-k:]+arr[:-k]
arr=[1,2,3,4,5]
obj=Solution()
print(obj.rotate_array(arr,2))
```
