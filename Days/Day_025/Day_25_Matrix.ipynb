# 49.Count Even Numbers
- arr = [[1, 2, 3], [4, 5, 6],[7, 8, 9]]

- output=4

```python
class Solution:
    def count_even(self,arr):
        count=0
        for row in arr:
            for value in row:
                if value %2==0:
                    count+=1
        return count
arr=[[1, 2, 3], 
       [4, 5, 6],
       [7, 8, 9]]
obj=Solution()
print(obj.count_even(arr))
```

# 50.Search an Element

- arr = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

output=True

```python
class Solution:
    def search(self,arr,target):
        for i in range(len(arr)):
            for j in range(len(arr[0])):
                if arr[i][j]==target:
                    return True
        return False
arr=[[1,2,3],
    [4,5,6],
    [7,8,9]]
obj=Solution()
print(obj.search(arr,5))
```
