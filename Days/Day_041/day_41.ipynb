# 69.Move Negative Numbers to One Side
- Input:  [1, -2, 3, -4, 5, -6]
- Output: [-2, -4, -6, 1, 5, 3]

```python
class Solution:
    def move_negativenumber(self,arr):
        left=0
        for i in range(len(arr)):
            if arr[i]<0:
                arr[left],arr[i]=arr[i],arr[left]
                left+=1
        return arr
arr=[1,-2,3,-4,5,-6]
obj=Solution()
print(obj.move_negativenumber(arr))
```

# 70.moveNegativesRight
- Input:[1, -2, 3, -4, 5, -6]

- Output:
[1, 3, 5, -4, -2, -6]

```python
class Solution:
    def moveNegativesRight(self, arr):
        left = 0
        right = len(arr) - 1
        while left <= right:
            if arr[left] < 0:
                arr[left], arr[right] = arr[right], arr[left]
                right -= 1
            else:
                left += 1
        return arr
arr = [1, -2, 3, -4, 5, -6]
obj = Solution()
print(obj.moveNegativesRight(arr))
```
