# 46. Print Matrix

```python
class Solution:
    def printamatrix(self,arr):
        for row in arr:
            print(row)
arr=[[1,2,3],
    [4,5,6],
    [7,8,9]]
obj=Solution()
print(obj.printamatrix(arr))
```

# 47.sum of matrix

```python
class Solution:
    def matrix_sum(self,arr):
        total=0
        for row in arr:
            for value in row:
                total+=value
        return total
arr=[[1,2,3],
    [4,5,6],
    [7,8,9]]
obj=Solution()
print(obj.matrix_sum(arr))
```
