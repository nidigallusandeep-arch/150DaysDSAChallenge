# 82.Common Elements in Two Arrays
- arr1 = [10, 20, 30, 40, 50]
- arr2 = [20, 40, 60, 80]
- output :- [20, 40]

```python
class Solution:
    def common_elements(self,arr1,arr2):
        result=[]
        for num in arr1:
            if num in arr2 and num not in result:
                result.append(num)
        return result
arr1=[10,20,30,40,50]
arr2=[20,40,60,80]
obj=Solution()
print(obj.common_elements(arr1,arr2))
```
