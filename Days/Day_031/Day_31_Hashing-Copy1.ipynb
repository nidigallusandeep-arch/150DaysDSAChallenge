# 58.Intersection of Two arrays with Duplicate Elements
- Input: a[] = [1, 2, 1, 3, 1], b[] = [3, 1, 3, 4, 1]
- Output: [1, 3]
- Explanation: 1 and 3 are the only common elements and we need to print only one occurrence of common elements.

```python
class Solution:
    def intersect(self, a, b):
        set_b = set(b)
        result = set()
        for num in a:
            if num in set_b:
                result.add(num)
        return list(result)
a = [1, 2, 1, 3, 1]
b = [3, 1, 3, 4, 1]
obj = Solution()
print(obj.intersect(a, b))
```
