# 53.Rotate Matrix 90° Clockwise
- Input
[[1 2 3],
[4 5 6],
[7 8 9]]

- Output
[[7 4 1],
[8 5 2],
[9 6 3]]

```python
class Solution:
    def rotate(self,arr):
        n=len(arr)
        for i in range(n):
            for j in range(1+1,n):
                arr[i][j], arr[j][i] = arr[j][i], arr[i][j]
            for row in arr:
                row.reverse()
            return arr
arr=[[1,2,3],
    [4,5,6],
    [7,8,9]]
obj=Solution()
print(obj.rotate(arr))
```

# 54.Spiral Matrix

# Input
[[1 2 3]
[4 5 ]
[7 8 9]]

- Output
1 2 3 6 9 8 7 4 5

```python
class Solution:
    def spiralOrder(self, arr):
        result = []
        top = 0
        bottom = len(arr) - 1
        left = 0
        right = len(arr[0]) - 1
        while top <= bottom and left <= right:
            # Left → Right
            for j in range(left, right + 1):
                result.append(arr[top][j])
            top += 1
            # Top → Bottom
            for i in range(top, bottom + 1):
                result.append(arr[i][right])
            right -= 1
            # Right → Left
            if top <= bottom:
                for j in range(right, left - 1, -1):
                    result.append(arr[bottom][j])
                bottom -= 1
            # Bottom → Top
            if left <= right:
                for i in range(bottom, top - 1, -1):
                    result.append(arr[i][left])
                left += 1
        return result
arr = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]
obj = Solution()
print(obj.spiralOrder(arr))
```
